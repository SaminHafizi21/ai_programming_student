```python
import random, string

TARGET = "hond"
LETTERS = string.ascii_uppercase

def random_word():
    return ''.join(random.choice(LETTERS) for _ in range(len(TARGET)))

def fitness(word):
    score = 0
    for i in range(len(TARGET)):
        if word[i] == TARGET[i]:
            score += 1
    return score            # (1) aantal letters op de juiste plaats

def crossover(p1, p2):
    point = random.randint(1, len(p1) - 1)
    return p1[:point] + p2[point:]            # (2)

def mutate(word, rate):
    chars = list(word)
    for i in range(len(chars)):
        if random.random() < rate:           # (3) met kans rate
            chars[i] = random.choice(LETTERS)  # (4) willekeurige letter
    return ''.join(chars)

population = [random_word() for _ in range(50)]
print(f"Pop: {population}")
for gen in range(200):
    population.sort(key=fitness, reverse=True)     # (5) beste eerst
    best = population[0]
    print(gen, best, fitness(best))
    if fitness(best) == fitness(TARGET):               # (6) stoppen als perfect
        break
    parents = population[:10]         # (7) de 10 beste
    children = []
    while len(children) < len(population):
        p1, p2 = random.sample(parents, 2)
        children.append(mutate(crossover(p1, p2), 0.1))  # (8) kind maken: crossover, dan mutatie
    population = children
```

**Bit Evolver (OneMax)**

*Probleem*

Een individu is een lijst van 20 bits, bv. [1, 0, 1, 1, 0, ...].

Doel: evolueer naar een lijst met alleen 1'en: [1, 1, 1, ..., 1].

Dit is het OneMax-probleem uit je cursus.


```python
"""
Oefening 1: Bit Evolver (Genetisch Algoritme - OneMax)
=======================================================
Evolueer een populatie van willekeurige bitlijsten naar een lijst met alleen 1'en.
"""
import random


class BitEvolver:
    def __init__(self, length=20, population_size=30, mutation_rate=0.05, num_generations=200):
        self.length = length
        self.population_size = population_size
        self.mutation_rate = mutation_rate
        self.num_generations = num_generations

    def generate_individual(self, length):
        # TODO stap 1: lijst van `length` willekeurige bits
        individual = []
        for i in range(length):
            individual.append(random.randint(0,1))
        return individual

    def calculate_fitness(self, individual):
        return sum(individual)

    def generate_population(self, size, length):
        # TODO stap 3: lijst van `size` individuen
        population = []
        for i in range(size):
            population.append(self.generate_individual(length))
        return population

    def crossover(self, parent1, parent2):
        # TODO stap 4: knippen en plakken
        point = random.randint(1, len(parent1)-1)
        return parent1[:point] + parent2[point:]

    def mutate(self, individual, rate):
        # TODO stap 5: kopie maken, elke bit omdraaien met kans `rate`
        new = individual.copy()
        for i in range(len(new)):
            if random.random() < rate:
                if new[i] == 0:
                    new[i] = 1
                else:
                    new[i] = 0
        return new
        
        

    def select_best(self, population):
        # TODO stap 6: sorteren, 10 beste teruggeven
        population.sort(key=self.calculate_fitness, reverse=True)
        return population[:10]
        

    def genetic_algorithm(self):
        # TODO stap 7
        population = self.generate_population(self.population_size, self.length)
        
        for gen in range(self.num_generations):
            
            #van de generatie lijst de beste eruit halen
            parents = self.select_best(population)
            bestOfParent = parents[0]
            
            if gen % 5 == 0:
                print(f"Generatie {gen}: {''.join(map(str, bestOfParent))} fitness {self.calculate_fitness(bestOfParent)}")
                
            
            #fitness vergelijken en stoppen als perfect   
            if self.calculate_fitness(bestOfParent) == self.length:
                print(f"Gevonden in generatie {gen}!")
                return bestOfParent
            
            # nieuwe generatie maken
            children = []
            while len(children) < self.population_size:
                p1, p2 = random.sample(parents, 2)
                child = self.crossover(p1, p2)
                child = self.mutate(child, self.mutation_rate)
                children.append(child)
            population = children
        return parents[0]

def main():
    evolver = BitEvolver()
    best = evolver.genetic_algorithm()
    print("Beste:", ''.join(map(str, best)))
    
        
if __name__ == "__main__":
    main()
```

**Woord Generator**

*Probleem*

Een individu is een zin van willekeurige hoofdletters en spaties, bv. "QWXA PLMZRTK".
Doel: evolueer naar de doelzin "HALLO WERELD".

```python
"""
Oefening 2: Woord Generator (Genetisch Algoritme)
==================================================
Evolueer een populatie van willekeurige zinnen naar een doelzin.
"""
import random
import string


class WordPuzzle:
    CHARACTERS = string.ascii_uppercase + " "    # A-Z en spatie

    def __init__(self, target_phrase, population_size=100, mutation_rate=0.02, num_generations=1000):
        self.target_phrase = target_phrase
        self.population_size = population_size
        self.mutation_rate = mutation_rate
        self.num_generations = num_generations

    def generate_phrase(self, length):
        # TODO stap 1
        random_phrase = ""
        for i in range(length):
            random_phrase += random.choice(self.CHARACTERS)
        return random_phrase

    def calculate_fitness(self, phrase):
        # TODO stap 2: juiste tekens / lengte
        score = 0
        
        for i in range(len(self.target_phrase)):
            if self.target_phrase[i] == phrase[i]:
                score += 1
        return score/len(self.target_phrase)
    
    def generate_population(self, size, phrase_length):
        # TODO stap 3
        population = []
        for i in range(size):
            population.append(self.generate_phrase(phrase_length))
        return population

    def crossover(self, p1, p2):
        # TODO stap 4
        point = random.randint(1, len(p1)-1)
        return p1[:point]+p2[point:]
        

    def mutate(self, phrase, rate):
        # TODO stap 5: string → lijst → aanpassen → string
        chars = list(phrase)
        for i in range(len(chars)):
            if random.random() < rate:
                chars[i] = random.choice(self.CHARACTERS)
        return ''.join(chars)

    def select_best(self, population):
        # TODO stap 6: beste 20%
        population.sort(key=self.calculate_fitness, reverse=True)
        return population[:20]

    def genetic_algorithm(self):
        # TODO stap 7
        population = self.generate_population(self.population_size, len(self.target_phrase) )
        for gen in range(self.num_generations):
            parents = self.select_best(population)
            best_parent = parents[0]
        
            if gen % 10 == 0:
                print(gen, best_parent, self.calculate_fitness(best_parent))
        
            if self.calculate_fitness(best_parent) == 1.0:
                return best_parent
            
            children = [] #nieuwe generatie
            while len(children) < self.population_size:
                p1, p2 = random.sample(parents, 2)
                child = self.crossover(p1, p2)
                child = self.mutate(child, self.mutation_rate)
                
                children.append(child)
                
            population = children
        return parents[0]


def main():
    puzzle = WordPuzzle("HALLO WERELD")
    best_phrase = puzzle.genetic_algorithm()
    print(f"Beste zin: {best_phrase}")


if __name__ == "__main__":
    main()
```

**Knapsack (zelf geschreven)**

Je gaat op kamp en je rugzak kan maximaal 15 kg dragen. Welke spullen neem je mee om de totale waarde zo hoog mogelijk te maken, zonder over 15 kg te gaan?

```python
"""
Oefening 3: Knapsack (Genetisch Algoritme, zonder library)
===========================================================
Kies voorwerpen met maximale waarde, zonder de capaciteit te overschrijden.
"""
import random


class Knapsack:
    def __init__(self, gewichten, waarden, capaciteit,
                 population_size=50, mutation_rate=0.05, num_generations=100):
        self.gewichten = gewichten
        self.waarden = waarden
        self.capaciteit = capaciteit
        self.population_size = population_size
        self.mutation_rate = mutation_rate
        self.num_generations = num_generations

    def generate_individual(self, length):
        # TODO stap 1
        individu = []
        for i in range(length):
            individu.append(random.randint(0, 1)) 
        
        return individu

    def total_weight(self, individual):
        # TODO stap 2: som van de gewichten waar de bit 1 is
        total = 0
        for i in range(len(individual)):
            if individual[i] == 1:
                total += self.gewichten[i]
        return total

    def total_value(self, individual):
        # TODO stap 3: som van de waarden waar de bit 1 is
        total = 0
        for i in range(len(individual)):
            if individual[i] == 1:
                total += self.waarden[i]
        return total

    def calculate_fitness(self, individual):
        # TODO stap 4: te zwaar → 0, anders de totale waarde
        
        if self.total_weight(individual) > self.capaciteit:
            return 0
        else:
            return self.total_value(individual)
    
    
    
    def generate_population(self, size, length):
        # TODO stap 5
        population = []
        for i in range(size):
            population.append(self.generate_individual(length))
        return population

    def crossover(self, p1, p2):
        # TODO stap 5
        point = random.randint(1, len(p1)-1)
        return p1[:point] + p2[point:]

    def mutate(self, individual, rate):
        # TODO stap 5
        bits = list(individual)
        for i in range(len(bits)):
            if random.random() < rate:
                if bits[i] == 1:
                    bits[i] = 0
                else:
                    bits[i] = 1
        return bits


    def select_best(self, population):
        # TODO stap 5: beste 20%
        population.sort(key=self.calculate_fitness, reverse=True)
        return population[:20]

    def genetic_algorithm(self):
        # TODO stap 6
        population = self.generate_population(self.population_size, len(self.gewichten))
        
        for gen in range(self.num_generations):
            parents = self.select_best(population)
            best_parent = parents[0]
            
            if gen % 10 ==0:
                print(f"Generatie {gen}: {best_parent} waarde {self.total_value(best_parent)} gewicht {self.total_weight(best_parent)}")

            children = []
            while len(children) < self.population_size:
                p1, p2 = random.sample(parents, 2)
                child = self.crossover(p1, p2)
                child = self.mutate(child, self.mutation_rate)
                children.append(child)
                
            population = children
        return self.select_best(population)[0]


def main():
    namen      = ["Laptop", "Tent", "Slaapzak", "Fototoestel", "Waterfles", "Boek", "Jas", "Powerbank"]
    gewichten  = [3, 7, 4, 2, 1, 2, 5, 1]
    waarden    = [30, 40, 25, 35, 10, 8, 20, 15]
    capaciteit = 15

    rugzak = Knapsack(gewichten, waarden, capaciteit)
    best = rugzak.genetic_algorithm()

    print("Beste:", best, "waarde", rugzak.total_value(best), "gewicht", rugzak.total_weight(best))
    print("Mee:", [namen[i] for i in range(len(best)) if best[i] == 1])


if __name__ == "__main__":
    main()
```
