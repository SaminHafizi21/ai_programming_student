### Voorbeeld 1
Minimaliseer $f(x) = \frac{1}{2}x^T P x + q^T x$ met beperking $l \leq Ax \leq u$ en met $$P =  \begin{bmatrix}
  4 & 1 \\
  1 & 2 \\
\end{bmatrix} , q = \begin{bmatrix}
  1 \\
  1 \\
\end{bmatrix} , A = \begin{bmatrix}
  1 & 1 \\
  -1 & 2 \\
  2 & 1 \\
\end{bmatrix}, l = \begin{bmatrix}
  1 \\
  0 \\
  2 \\
\end{bmatrix}, u = \begin{bmatrix}
  2 \\
  2 \\
  3 \\
\end{bmatrix}$$

```python
import numpy as np
from scipy.sparse import csc_matrix #CSC staat voor compact sparse column. Dit is wat OSQP verwacht te krijgen
import osqp

# Matrices opstellen / ingeven
P = np.array([[4, 1], [1, 2]])
q = np.array([1, 1])
A = np.array([[1, 1], [-1, 2], [2, 1]])
l = np.array([1, 0, 2])
u = np.array([2, 2, 3])

# P en A moeten als sparse matrices worden doorgegeven
sparse_A = csc_matrix(A)
sparse_P = csc_matrix(P)

# OSQP initialisatie
prob = osqp.OSQP()

# setup
prob.setup(sparse_P, q, sparse_A, l, u)

# Oplossing
result = prob.solve()

print("Oplossing:")
print(result.x)
```

### Voorbeeld 2

We willen $c^Tx$ minimaliseren, met randvoorwaarde $Gx \leq h$. We moeten nu zelf nog een beetje werk doen om dit om te zetten naar de formulering hierboven. Concreet is $$c =  \begin{bmatrix}
  1 \\
  -1 \\
\end{bmatrix} , G = \begin{bmatrix}
  1 & -1 \\
  -1 & -2 \\
  2 & 1 \\
\end{bmatrix}, h = \begin{bmatrix}
  -1 \\
  -2 \\
  4 \\
\end{bmatrix}$$

Als je goed kijkt zie je dat $h$ de bovengrens is. Er is geen ondergrens gegeven, dus we zullen een ondergrens maken door een array met $-\infty$ te vullen. $G$ is gewoon onze $A$, dus dat is makkelijk.

$c$ is het lineaire stuk van de te minimaliseren kostfunctie, dus $c = q$. Bij dit probleem is er geen kwadratisch stuk, dus voor de matrix $P$ gaan we een vierkante 0-matrix meegeven.

```python
import numpy as np
import osqp

# gegeven
P = np.array([[0.0, 0.0], [0.0, 0.0]])
q = np.array([1.0, -1.0])
A = np.array([[1.0, -1.0], [-1.0, -2.0], [2.0, 1.0]])
l = np.array([-np.inf, -np.inf, -np.inf])
u = np.array([-1.0, -2.0, 4.0])

# P en A moeten als sparse matrices worden doorgegeven
sparse_P = csc_matrix(P)
sparse_A = csc_matrix(A)

# OSQP initialisatie
prob = osqp.OSQP()

# Setup van het probleem met de gedefinieerde matrices
prob.setup(sparse_P, q, sparse_A, l, u, verbose=False)

# Oplossing berekenen
result = prob.solve()

# --- Resultaten tonen ---
if result.info.status == 'solved':
    optimal_x = result.x
    optimal_value = result.info.obj_val

    print(f"De status van de oplossing is: {result.info.status}")
    print(f"De optimale waarde voor x is: {optimal_x}")
    print(f"De minimale waarde van de doelfunctie (c^T * x) is: {optimal_value}")
else:
    print(f"Er kon geen optimale oplossing worden gevonden. Status: {result.info.status}")
```

Een financieel analyst wil haar aandelenportefeuille optimaliseren. Ze wil dus haar budget verdelen over verschillende aandelen, in de hoop om de verwachte return van die aandelen te maximaliseren.

*Doel*: maximaliseer de verwachte return $\sum_{i=1}^n w_i \cdot r_i$
Hierbij is $w_i$ het gewicht, wat we dus zo optimaal willen zoeken, en $r_i$ de verwachte return van aandeel $i$.

*Randvoorwaarden*
- De som van alle gewichtjes moet in totaal 1 zijn: $\sum_{i=1}^n w_i = 1$
- De analyst wil bovendien haar risico beperken. Hiervoor krijg je een covariance matrix tussen alle returns, die dus op rij $i$ en kolom $j$ aangeeft wat het verband is tussen die twee aandelen.

Gegeven data:

```python
import numpy as np

# aantal aandelen
num_stocks = 3

# Verwachte return (R)
np.random.seed(42)
expected_returns = np.random.uniform(0.05, 0.15, num_stocks) # hier ben je optimistisch over de beurssituatie

# correlatie matrix(Sigma)
correlation_matrix = np.array([[1.0, 0.5, -0.3],
                               [0.5, 1.0, 0.2],
                               [-0.3, 0.2, 1.0]])

# correlation matrix omzetten naar covariance matrix
volatility = np.random.uniform(0.1, 0.3, num_stocks) # in financiële sector wordt volatiliteit gebruikt als synoniem voor de standaardafwijking op aandelen
covariance_matrix = np.outer(volatility, volatility) * correlation_matrix

print("Verwachte Returns (R):")
print(expected_returns)

print("\nCovariance Matrix (Sigma):")
print(covariance_matrix)
```

# Optimalisatie in een fabriek

We bekijken een probleem in een fabriek, waar een bedrijf zoveel mogelijk winst wil maken met 2 producten die ze kan maken. Er worden twee producten gemaakt: een product AquaSparkle, en een product BioBurst. Om deze producten te maken, heb je twee grondstoffen nodig (X en Y). Elk product heeft een bepaalde winst per verkochte eenheid, en de grondstoffenvoorraad is beperkt:
- Product AquaSparkle vereist 2 eenheden grondstof X, 1 eenheid grondstof Y
- Product BioBurst vereist 1 eenheid grondstof X, 3 eenheden grondstof Y
- Er zijn 100 grondstoffen X
- Er zijn 90 grondstoffen Y
De winst voor een verkocht AquaSparkle is 5 euro. De winst voor BioBurst is 8 euro. 

Bepaal het optimale productieproces: hoeveel AquaSparkles en hoeveel Biobursts moeten er worden gemaakt ?


### Code omzetten naar .py file

Zet je uiteindelijke code om in wrappercode. (basisversie)

Zorg in de wrappercode dat je bij initialisatie de volgende waardes kan aanpassen: (versie met parameters)
- de beperking op het aantal grondstoffen X
- de beperking op het aantal grondstoffen Y
- de winst van A
- de winst van B
- (de verhouding van grondstoffen voor de producten mag je constant houden en dus hardcoden)


```python
# production_optimizer.py
import numpy as np
from scipy.sparse import csc_matrix
import osqp

class ProductionOptimizer:
    """
    Een wrapper om een productieprobleem met 2 producten en 2 grondstoffen
    te optimaliseren met behulp van de OSQP-solver.
    """
    def __init__(self, limit_x, limit_y, profit_a, profit_b):
        """
        Initialiseert de optimizer met de probleemparameters.

        Args:
            limit_x (float): De maximale voorraad van grondstof X.
            limit_y (float): De maximale voorraad van grondstof Y.
            profit_a (float): De winst per eenheid van product A (AquaSparkle).
            profit_b (float): De winst per eenheid van product B (BioBurst).
        """
        self.limit_x = limit_x
        self.limit_y = limit_y
        self.profit_a = profit_a
        self.profit_b = profit_b
        
        # Hardgecodeerde verhoudingen van grondstoffen
        # Product A: 2x X, 1x Y
        # Product B: 1x X, 3x Y
        self.A_coeffs = [2, 1]
        self.B_coeffs = [1, 3]

        self.result = None

    def solve(self):
        """
        Stelt het probleem op en lost het op met OSQP.
        """
        # P is altijd een nulmatrix voor dit lineaire probleem
        P = csc_matrix((2, 2))
        
        # q bevat de negatieve winstcoëfficiënten
        q = np.array([-self.profit_a, -self.profit_b])
        
        # A bevat de coëfficiënten van de randvoorwaarden
        A = csc_matrix([
            [self.A_coeffs[0], self.B_coeffs[0]],  # Grondstof X
            [self.A_coeffs[1], self.B_coeffs[1]],  # Grondstof Y
            [1, 0],                               # x_A >= 0
            [0, 1]                                # x_B >= 0
        ])
        
        # l en u definiëren de grenzen
        l = np.array([-np.inf, -np.inf, 0, 0])
        u = np.array([self.limit_x, self.limit_y, np.inf, np.inf])
        
        # OSQP setup en solve
        prob = osqp.OSQP()
        prob.setup(P, q, A, l, u, verbose=False)
        self.result = prob.solve()
        
        return self.result.info.status == 'solved'

    def print_solution(self):
        """
        Print de gevonden oplossing op een leesbare manier.
        """
        if self.result and self.result.info.status == 'solved':
            aqua_sparkle_units, bio_burst_units = self.result.x
            max_profit = -self.result.info.obj_val
            
            print("--- Optimale Productie (Wrapper-versie) ---")
            print(f"Parameters: Limiet X={self.limit_x}, Limiet Y={self.limit_y}, Winst A={self.profit_a}, Winst B={self.profit_b}")
            print("-" * 20)
            print(f"Aantal AquaSparkle te produceren: {aqua_sparkle_units:.0f}")
            print(f"Aantal BioBurst te produceren: {bio_burst_units:.0f}")
            print(f"Maximale winst: € {max_profit:.2f}")
        else:
            status = self.result.info.status if self.result else "Nog niet opgelost"
            print(f"Optimalisatie mislukt of nog niet uitgevoerd. Status: {status}")


if __name__ == '__main__':
    # Initialiseer de optimizer met de waarden uit de opgave
    optimizer = ProductionOptimizer(limit_x=100, limit_y=90, profit_a=5, profit_b=8)
    
    # Los het probleem op
    success = optimizer.solve()
    
    # Toon de resultaten
    if success:
        optimizer.print_solution()

```
