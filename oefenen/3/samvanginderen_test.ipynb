```python
import numpy as np
from scipy.sparse import csc_matrix #CSR staat voor compact sparse column. Dit is wat OSQP verwacht te krijgen
import osqp
```

matrices bepalen

M = aantal uren montage; max 120u
I = aantal uren inspectie; max 80u
S = smartphone
T = tablet

winst S = 150
winst T = 200

1S = 2M + 1I
1T = 1T + 2I

aantal smartphones en tablets berekenen voor max winst met gegeven uren
dit is de winst functie q:
--> S = x, T = y
q = 150x+200y
--> q = [150, 200]

grenswaarden bepalen
--> 0 <= 2x+y <= 120
0 <= x+2y <= 80
0 <= x <= inf
0 <= y <= inf

u = [120, 80, inf, inf]
l = [0, 0, 0, 0]
A = [ [2, 1] , [1, 2] , [1, 0] , [0, 1] ]
we hebben geen kwadratische kost dus P is een nul matrix 
P = [ [0, 0] , [0, 0] ]

```python
# Matrices opstellen / ingeven
P = np.array([[0, 0], [0, 0]])
q = np.array([150, 200])
A = np.array([[2, 1], [1, 2], [1, 0], [0, 1]])
l = np.array([0, 0, 0, 0])
u = np.array([120, 80, np.inf, np.inf])

q = -q
# P en A moeten als sparse matrices worden doorgegeven
sparse_A = csc_matrix(A)
sparse_P = csc_matrix(P)

# OSQP initialisatie
prob = osqp.OSQP()

# setup
prob.setup(sparse_P, q, sparse_A, l, u)

# Oplossing
result = prob.solve()

print("Oplossing:")
print(result.x)
```
