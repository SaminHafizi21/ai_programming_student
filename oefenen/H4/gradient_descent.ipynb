Een koerier vertrekt in Antwerpen, moet langs 5 steden en komt daarna terug naar Antwerpen. Zoek een zo kort mogelijke route. Er zijn te veel routes om ze allemaal te proberen, en een greedy aanpak blijft in een lokaal minimum steken. Vul de code aan.

```python
import math, random

steden = {"Antwerpen": (0, 0), "Gent": (-5, -1), "Brussel": (-1, -4),
          "Leuven": (2, -4), "Hasselt": (6, -3), "Brugge": (-9, 0)}

def afstand(a, b):
    (x1, y1), (x2, y2) = steden[a], steden[b]
    return math.sqrt((x1 - x2)**2 + (y1 - y2)**2)

def kost(route):
    totaal = 0
    for i in range(len(route)):
        totaal += afstand(route[i], route[(i+1) % len(route)])  # (2) volgende stad, laatste → terug naar eerste
    return totaal

def random_neighbor(route):
    nieuw = route.copy()
    i, j = random.sample(range(len(nieuw)), 2)
    nieuw[i], nieuw[j] = nieuw[j], nieuw[i]                 # (3) wissel twee steden
    return nieuw

def simulated_annealing(initial_state, cost_function, kmax):
    s = initial_state
    s_best = s
    for k in range(kmax):
        T = max(1 - (k + 1)/kmax, 1e-3)         
        s_new = random_neighbor(s)
        if cost_function(s_new) < cost_function(s):
            s = s_new
        else:
            delta = cost_function(s_new) - cost_function(s)  # (5)
            P = math.exp(-delta/T)      # (6)
            if random.random() < P:      # (7)
                s = s_new
        if cost_function(s) < cost_function(s_best):          # (8)
            s_best = s
    return s_best
```
