```python
"""
Oefening H4-1: Gradient Descent
================================
Zoek het minimum van f(x) = (x - 3)^2 + 2 door de helling naar beneden te volgen.
"""
class GradientDescent:
    def __init__(self, initial_x=0.0, learning_rate=0.1,convergence_threshold=0.0001, max_steps=1000):
        self.initial_x = initial_x
        self.learning_rate = learning_rate
        self.convergence_threshold = convergence_threshold
        self.max_steps = max_steps

    def cost_function(self, x):
        return (x-3)**2 + 2

    def gradient(self, x):
        return 2*x - 6
        
    def run(self):
        # TODO stap 5
        x = self.initial_x
        for k in range(self.max_steps):
            grad = self.gradient(x)
            
            new_x = x - self.learning_rate * grad
            
            if k%5 == 0:
                print(f"Stap {k}: x = {x:.4f}, kost = {self.cost_function(x):.4f}")
            
            if abs(new_x - x) < self.convergence_threshold:
                return new_x 
                
            x = new_x
        return x


def main():
    gd = GradientDescent()
    resultaat = gd.run()
    print(f"Minimum bij x = {resultaat:.4f}, kost = {gd.cost_function(resultaat):.4f}")


if __name__ == "__main__":
    main()
```

**Oefening 2 — Routecalculator**    

Een koerier vertrekt in Antwerpen, rijdt langs Brussel, Gent en Leuven, en komt terug naar Antwerpen. Later zoek je met simulated annealing de kortste volgorde. In deze oefening maak je eerst de gereedschappen die daarvoor nodig zijn. Er zit nog geen algoritme in.

```python
"""
Oefening 2: Routecalculator
===============================
Werken met een afstandsmatrix: afstand van een tour en buren maken.
"""
import random

import numpy as np


class RouteCalculator:
    def __init__(self, distance_matrix, city_names):
        self.distance_matrix = distance_matrix
        self.city_names = city_names
        self.num_cities = len(distance_matrix)

    def distance(self, a, b):
        # TODO stap 1: afstand van stad a naar stad b
        return self.distance_matrix[a][b]
        
    def total_distance(self, tour):
        total_distance = 0
        for i in range(len(tour)):
            van = tour[i]
            naar = tour[(i+1) % len(tour)]
            total_distance += self.distance(van, naar)
            
        return total_distance

    def random_tour(self):
        # TODO stap 3: alle steden in willekeurige volgorde
        tour = list(range(self.num_cities))
        random.shuffle(tour)
        return tour
        

    def swap_two(self, tour):
        # TODO stap 4: kopie, 2 posities wisselen
        copy_tour = tour.copy()
        i, j = random.sample(range(len(copy_tour)), 2)
        
        copy_tour[i], copy_tour[j] = copy_tour[j], copy_tour[j]
        
        return copy_tour
    
    def tour_names(self, tour):
        # TODO stap 5: indexen → namen
        namen = []
        for i in range(len(tour)):
            namen.append(self.city_names[i])
        return namen


def main():
    distance_matrix = np.array([
        [0, 29, 20, 21],
        [29, 0, 15, 18],
        [20, 15, 0, 25],
        [21, 18, 25, 0]
    ])
    namen = ["Antwerpen", "Brussel", "Gent", "Leuven"]
    calc = RouteCalculator(distance_matrix, namen)

    for tour in ([0, 1, 2, 3], 
                 [0, 2, 1, 3], 
                 [0, 1, 3, 2]):
        print(tour, calc.total_distance(tour), calc.tour_names(tour))

    print("Brussel → Leuven:", calc.distance(1, 3))

    tour = [0, 1, 2, 3]
    print("Voor de swap:", tour)
    print("Na de swap:  ", calc.swap_two(tour))
    print("Origineel nog:", tour)

    print("Willekeurige tour:", calc.random_tour())


if __name__ == "__main__":
    main()
```
